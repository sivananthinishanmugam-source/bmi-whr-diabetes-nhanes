```python
from pathlib import Path

# The notebooks are stored in <package>/notebooks.
notebooks_dir = Path.cwd()
if notebooks_dir.name != "notebooks":
    candidate = notebooks_dir / "notebooks"
    if candidate.exists():
        notebooks_dir = candidate

project_root = notebooks_dir.parent
nhanes_dir = project_root / "src" / "nhanes"
jordan_dir = project_root / "src" / "jordan_external_validation"
expected_nhanes = project_root / "outputs" / "nhanes"
expected_jordan = project_root / "outputs" / "jordan"

assert (project_root / "README.md").exists(), project_root
print("Project root:", project_root)
```

```python
import subprocess
import sys

DOWNLOAD_NHANES = False  # Set True to download missing public CDC XPORT files.

raw_dir = project_root / "data" / "raw"
existing = sorted(raw_dir.glob("*.xpt"))
print("Existing NHANES XPT files:", len(existing))

if DOWNLOAD_NHANES:
    subprocess.run([sys.executable, "download_sources.py"], cwd=nhanes_dir, check=True)
    existing = sorted(raw_dir.glob("*.xpt"))

print("NHANES XPT files after optional download:", len(existing))
```
